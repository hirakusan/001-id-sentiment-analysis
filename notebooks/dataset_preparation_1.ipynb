```python
import sys, os
sys.path.append(os.path.abspath("../src"))

from data.dataset_loader import HuggingFaceDatasetLoader
```

```python
loader = HuggingFaceDatasetLoader("adealvii/Cleaned-Indonesian-Tweet")
df = loader.load_dataset()
```

```python
if df is not None:
    df['label'] = df['full_text'].apply(lambda x: 'positif' if 'baik' in str(x).lower() else 'negatif')
    loader.df = df
else:
    print("❌ Gagal memuat dataset.")
```

```python
info = loader.explore_dataset()
for key, value in info.items():
    print(f"\n{key.upper()}:" if isinstance(value, dict) else f"{key}: {value}")
    if isinstance(value, dict):
        for k, v in value.items():
            print(f"  {k}: {v}")
```

```python
from sklearn.model_selection import train_test_split
train_df, test_df = loader.prepare_data(
    text_column="full_text",
    label_column="label",
    test_size=0.2,
    stratify=True,
    random_state=42
)
```
